## Key Insights

1. **Class imbalance** — ~27% churn rate, use ROC-AUC as primary metric
2. **Contract** — month-to-month contracts churn at ~43%, two-year at ~11%
3. **Tenure** — recent customers churn far more frequently
4. **InternetService** — fiber optic users churn more than DSL
5. **PaymentMethod** — electronic check users have highest churn
6. **MonthlyCharges** — churners trend toward higher monthly charges
